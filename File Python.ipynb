## Creazione Dataset unico
In order to create a one single dataset what i will do is to transform the unique vaalues of each table into attributes. this will be done only for Motivo Contatto and Tipo Violenza

### Pivot Table di motivi

```python
df_wide = df_motivo.pivot_table(
    index=['REF_AREA', 'Territorio', 'Sesso', 'TIME_PERIOD'],  # cosa identifica una riga
    columns='Motivi della chiamata',                            # i 16 valori → 16 colonne
    values= 'MC_Osservazione',                                   # il valore nelle celle
    aggfunc='sum',                                              # se ci sono duplicati li somma
    fill_value=0                                                # 0 dove una combinazione manca
).reset_index()

df_wide.columns.name = None   # toglie l'etichetta "Motivi della chiamata" sopra le colonne
df_wide.head(5)
```

### Pivot Table per tpo di violenzA

```python
df.head(1)
```

```python
df_wide_violenze = df.pivot_table(
    index=['REF_AREA', 'Territorio', 'Sesso', 'TIME_PERIOD'],  # cosa identifica una riga
    columns='Tipo di violenza',                            # i 16 valori → 16 colonne
    values= 'TV_Osservazione',                                   # il valore nelle celle
    aggfunc='sum',                                              # se ci sono duplicati li somma
    fill_value=0                                                # 0 dove una combinazione manca
).reset_index()

df_wide_violenze.columns.name = None   # toglie l'etichetta "Tipo di violenza" sopra le colonne
df_wide_violenze.head(5)
```

### Drop Non disponibile

```python
df_wide_violenze.drop(columns=["Non disponibile"], inplace = True)
```

### merge 2 dataset

```python
big_dataset = pd.merge(df_wide, df_wide_violenze, how='outer', on=['REF_AREA', 'Territorio', 'Sesso', 'TIME_PERIOD'])

big_dataset.head()
```

```python
### download big dataset in csv
big_dataset.to_csv('Big Dataset', index=False)
```
